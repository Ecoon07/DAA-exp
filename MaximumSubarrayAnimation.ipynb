```python
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

arr = [-2, 1, -3, 4, -1, 2, 1, -5]

steps = [
    {
        "title": "Step 1: Original Array",
        "message": "Find the contiguous subarray having the maximum sum.",
        "highlight": []
    },
    {
        "title": "Step 2: Divide",
        "message": "Divide the array into left and right halves.",
        "highlight": [3, 4]
    },
    {
        "title": "Step 3: Solve Left Half",
        "message": "Maximum subarray in the left half is [4]. Sum = 4",
        "highlight": [3]
    },
    {
        "title": "Step 4: Solve Right Half",
        "message": "Maximum subarray in the right half is [2, 1]. Sum = 3",
        "highlight": [5, 6]
    },
    {
        "title": "Step 5: Crossing Subarray",
        "message": "Crossing subarray is [4, -1, 2, 1]. Sum = 6",
        "highlight": [3, 4, 5, 6]
    },
    {
        "title": "Step 6: Compare",
        "message": "Left = 4, Right = 3, Crossing = 6. Maximum = 6",
        "highlight": [3, 4, 5, 6]
    },
    {
        "title": "Final Result",
        "message": "Maximum Subarray = [4, -1, 2, 1]     Maximum Sum = 6",
        "highlight": [3, 4, 5, 6]
    }
]

fig, ax = plt.subplots(figsize=(10, 5))

def show_step(frame):
    ax.clear()

    step = steps[frame]

    colors = []
    for i in range(len(arr)):
        if i in step["highlight"]:
            colors.append("lightgreen")
        else:
            colors.append("lightblue")

    bars = ax.bar(
        range(len(arr)),
        arr,
        color=colors,
        edgecolor="black"
    )

    for i, bar in enumerate(bars):
        if arr[i] >= 0:
            y = arr[i] + 0.25
        else:
            y = arr[i] - 0.55

        ax.text(
            bar.get_x() + bar.get_width()/2,
            y,
            str(arr[i]),
            ha="center",
            fontsize=12,
            fontweight="bold"
        )

    ax.axhline(0, color="black", linewidth=1)

    ax.set_xticks(range(len(arr)))
    ax.set_xticklabels(
        ["A[0]", "A[1]", "A[2]", "A[3]",
         "A[4]", "A[5]", "A[6]", "A[7]"]
    )

    ax.set_ylim(-7, 7)

    ax.set_title(
        step["title"],
        fontsize=16,
        fontweight="bold"
    )

    ax.text(
        0.5,
        -0.17,
        step["message"],
        transform=ax.transAxes,
        ha="center",
        fontsize=11
    )

    if frame == 1:
        ax.axvline(
            3.5,
            linestyle="--",
            linewidth=2
        )

        ax.text(
            1.5,
            6,
            "LEFT HALF",
            ha="center",
            fontweight="bold"
        )

        ax.text(
            5.5,
            6,
            "RIGHT HALF",
            ha="center",
            fontweight="bold"
        )

    ax.set_ylabel("Array Value")

    plt.tight_layout()


animation = FuncAnimation(
    fig,
    show_step,
    frames=len(steps),
    interval=1800,
    repeat=True
)

from IPython.display import HTML
HTML(animation.to_jshtml())
```

```python
for i, frame_number in enumerate([0, 1, 4, 6], start=1):
    show_step(frame_number)

    plt.savefig(
        f"frame{i}.png",
        dpi=200,
        bbox_inches="tight"
    )

print("4 screenshots created successfully.")
```
